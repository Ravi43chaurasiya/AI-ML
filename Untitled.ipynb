<code>The parameter squeeze=True is deprecated and no longer works as expected.

So instead of using squeeze=True, do this:

import pandas as pd

df4 = pd.read_csv("aug_train.csv", usecols=["enrollee_id"])
df4 = df4["enrollee_id"]  # convert column to Series

print(type(df4))  # <class 'pandas.core.series.Series'>

Explanation

usecols=["enrollee_id"] → reads only that column.

df4["enrollee_id"] → extracts that column as a Series.</code>

## converter ##

```python
def rename(name):
    if name=="Royal Challengers Bangalore":
        return "RCB"

    else:
        return name

rename("Royal Challengers Bangalore")
```

```python
pd.read_csv("IPL.csv",converters={"bowling_team":rename})
```
